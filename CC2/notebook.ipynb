## 2. Obtención de Estados de Bell

### Generación Teórica de los Estados de Bell

Los estados de Bell son cuatro estados cuánticos específicos de dos cúbits que representan el **entrelazamiento máximo** posible. Matemáticamente, forman una base ortonormal para el espacio de Hilbert de un sistema de dos niveles y dos partículas ($C^2 \otimes C^2$).

Los cuatro estados se definen como:

*   **$\lvert\Phi^+\rangle = \frac{1}{\sqrt{2}}(\lvert00\rangle + \lvert11\rangle)$**
*   **$\lvert\Phi^-\rangle = \frac{1}{\sqrt{2}}(\lvert00\rangle - \lvert11\rangle)$**
*   **$\lvert\Psi^+\rangle = \frac{1}{\sqrt{2}}(\lvert01\rangle + \lvert10\rangle)$**
*   **$\lvert\Psi^-\rangle = \frac{1}{\sqrt{2}}(\lvert01\rangle - \lvert10\rangle)$**

### Construcción de circuitos

Partiendo siempre del estado fundamental del sistema $\lvert00\rangle$, la generación de cualquier estado de Bell requiere una receta de dos pasos fundamentales:

1.  **Creación de Superposición:** Se aplica una puerta **Hadamard ($H$)** sobre el primer cúbit (cúbit de control, índice `0`). Esto rompe el determinismo clásico del cúbit, transformando el estado $\lvert0\rangle$ en una combinación equitativa de ambos estados: $\frac{1}{\sqrt{2}}(\lvert0\rangle + \lvert1\rangle)$.
2.  **Entrelazamiento Coherente:** Se aplica una puerta **NOT Controlada ($CNOT$ o $CX$)**, utilizando el cúbit `0` como control y el cúbit `1` como objetivo (*target*). Dado que el control se encuentra en una superposición, la puerta arrastra mecánicamente al segundo cúbit a una correlación absoluta con el primero.

Para obtener las variaciones de signo (fase) o las inversiones de bit que diferencian a los cuatro estados, se intercalan puertas de inversión **Pauli-X ($X$)** antes o después del núcleo del circuito, alterando el estado inicial del sistema. Los circuitos para conseguir cada estado partiendo de la base estándar $\lvert00\rangle$ son los siguientes:

#### Estado $\lvert\Phi^+\rangle = \frac{1}{\sqrt{2}}(\lvert00\rangle + \lvert11\rangle)$
Es el estado de Bell más sencillo e ideal para empezar. No requiere ninguna preparación previa de los cúbits de entrada.

*   **Paso 1:** Se aplica la puerta **Hadamard ($H$)** al cúbit 0 (primer cúbit). El cúbit 0 pasa a superposición $\frac{1}{\sqrt{2}}(\lvert0\rangle + \lvert1\rangle)$, obteniendo como estado global del sistema:
    $$\frac{1}{\sqrt{2}}(\lvert00\rangle + \lvert10\rangle)$$
*   **Paso 2:** Se aplica la puerta **CNOT ($CX$)** usando el cúbit 0 como control y el cúbit 1 como objetivo. Esto entrelaza los estados de ambos cúbits de la siguiente manera:
    *   La componente $\lvert00\rangle$ se mantiene igual debido a que el cúbit de control (cúbit 0) está en el estado $\lvert0\rangle$.
    *   La componente $\lvert10\rangle$ se transforma en $\lvert11\rangle$, ya que el cúbit de control está en el estado $\lvert1\rangle$, el cúbit 1 pasa de $\lvert0\rangle$ a $\lvert1\rangle$.

#### Estado $\lvert\Phi^-\rangle = \frac{1}{\sqrt{2}}(\lvert00\rangle - \lvert11\rangle)$
Este estado introduce el signo menos (la fase relativa). Para conseguirlo, se necesita invertir el estado del primer cúbit antes de aplicar la superposición.

*   **Paso 1:** Se aplica una puerta **Pauli-X ($X$)** al cúbit 0 para transformarlo de $\lvert0\rangle$ a $\lvert1\rangle$. El estado global inicial pasa a ser $\lvert10\rangle$.
*   **Paso 2:** Se aplica la puerta **Hadamard ($H$)** al cúbit 0. Debido a que el cúbit arrancaba en el estado $\lvert1\rangle$, la rotación genera una superposición con signo negativo: $\frac{1}{\sqrt{2}}(\lvert0\rangle - \lvert1\rangle)$. Al combinarlo con el cúbit 1, el estado global del sistema pasa a ser:
    $$\frac{1}{\sqrt{2}}(\lvert00\rangle - \lvert10\rangle)$$
*   **Paso 3:** Se aplica la puerta **CNOT ($CX$)** usando el cúbit 0 como control y el cúbit 1 como objetivo:
    *   La componente $\lvert00\rangle$ se mantiene igual debido a que el cúbit de control (cúbit 0) está en el estado $\lvert0\rangle$.
    *   La componente $-\lvert10\rangle$ se transforma en $-\lvert11\rangle$, ya que el cúbit de control está en el estado $\lvert1\rangle$, arrastrando el signo menos y modificando el cúbit 1 de $\lvert0\rangle$ a $\lvert1\rangle$.

---

#### Estado $\lvert\Psi^+\rangle = \frac{1}{\sqrt{2}}(\lvert01\rangle + \lvert10\rangle)$
Este estado correlaciona los cúbits de forma inversa (si uno es 0, el otro es 1). Para obtenerlo, se altera el cúbit objetivo tras realizar el entrelazamiento base.

*   **Paso 1:** Se aplica la puerta **Hadamard ($H$)** al cúbit 0. El cúbit 0 pasa a superposición, obteniendo como estado global del sistema:
    $$\frac{1}{\sqrt{2}}(\lvert00\rangle + \lvert10\rangle)$$
*   **Paso 2:** Se aplica la puerta **CNOT ($CX$)** usando el cúbit 0 como control y el cúbit 1 como objetivo, dejando el sistema en el estado entrelazado base:
    $$\frac{1}{\sqrt{2}}(\lvert00\rangle + \lvert11\rangle)$$
*   **Paso 3:** Se aplica una puerta **Pauli-X ($X$)** en el cúbit 1 (segundo cúbit). Esta puerta invierte de forma determinista el valor del segundo cúbit en todas las componentes del sistema:
    *   La componente $\lvert00\rangle$ pasa a ser $\lvert01\rangle$ (el segundo cúbit cambia de 0 a 1).
    *   La componente $\lvert11\rangle$ pasa a ser $\lvert10\rangle$ (el segundo cúbit cambia de 1 a 0).

---

#### Estado $\lvert\Psi^-\rangle = \frac{1}{\sqrt{2}}(\lvert01\rangle - \lvert10\rangle)$
Eeste estado combina tanto la correlación inversa de los bits como el desfase negativo. Unimos las técnicas lógicas de los estados anteriores.

*   **Paso 1:** Se aplica una puerta **Pauli-X ($X$)** al cúbit 0 para transformarlo de $\lvert0\rangle$ a $\lvert1\rangle$, dejando el sistema en el estado inicial $\lvert10\rangle$.
*   **Paso 2:** Se aplica la puerta **Hadamard ($H$)** al cúbit 0, inyectando la fase negativa debido al estado inicial $\lvert1\rangle$. El estado global del sistema pasa a ser:
    $$\frac{1}{\sqrt{2}}(\lvert00\rangle - \lvert10\rangle)$$
*   **Paso 3:** Se aplica la puerta **CNOT ($CX$)** (Control: cúbit 0, Target: cúbit 1) para entrelazar las componentes:
    *   La componente $\lvert00\rangle$ se mantiene igual.
    *   La componente $-\lvert10\rangle$ se transforma en $-\lvert11\rangle$. El sistema queda temporalmente en el estado $\frac{1}{\sqrt{2}}(\lvert00\rangle - \lvert11\rangle)$.
*   **Paso 4:** Se aplica una puerta **Pauli-X ($X$)** en el cúbit 1 (segundo cúbit) para cruzar los valores finales:
    *   La componente $\lvert00\rangle$ se transforma en $\lvert01\rangle$.
    *   La componente $-\lvert11\rangle$ se transforma en $-\lvert10\rangle$.

```python
from qiskit import QuantumCircuit

# Diccionario para almacenar los 4 circuitos de los estados de Bell
bell_circuits = {}

# 1. Estado |Phi+> = (|00> + |11>) / sqrt(2)
qc_phi_plus = QuantumCircuit(2, 2) # Se parte de un circuito con estado inicial |00> y dos bits clásicos para la posterior medición
qc_phi_plus.h(0) # Se aplica la puerta H al qbit 0
qc_phi_plus.cx(0, 1) # Se aplica puerta CNOT con control 0 y target 1 
qc_phi_plus.measure([0, 1], [0, 1]) # Se realiza la medición de los qubits y se almacena la informacion en los bits clásicos
bell_circuits["phi+"] = qc_phi_plus # Se almacena el circuito en el diccionario

# 2. Estado |Phi-> = (|00> - |11>) / sqrt(2)
qc_phi_minus = QuantumCircuit(2, 2) # Se parte de un circuito con estado inicial |00> y dos bits clásicos para la posterior medición
qc_phi_minus.x(0) # Se empieza en |10> para que tras la H y CNOT cambie el signo
qc_phi_minus.h(0) # Se aplica la puerta H al qbit 0
qc_phi_minus.cx(0, 1) # Se aplica puerta CNOT con control 0 y target 1 
qc_phi_minus.measure([0, 1], [0, 1]) # Se realiza la medición de los qubits y se almacena la informacion en los bits clásicos
bell_circuits["phi-"] = qc_phi_minus # Se almacena el circuito en el diccionario

# 3. Estado |Psi+> = (|01> + |10>) / sqrt(2)
qc_psi_plus = QuantumCircuit(2, 2) # Se parte de un circuito con estado inicial |00> y dos bits clásicos para la posterior medición
qc_psi_plus.h(0) # Se aplica la puerta H al qbit 0
qc_psi_plus.cx(0, 1) # Se aplica puerta CNOT con control 0 y target 1 
qc_psi_plus.x(1) # Se aplica la puerta X en el qbit 1 para invertir sus valores lógicos
qc_psi_plus.measure([0, 1], [0, 1]) # Se realiza la medición de los qubits y se almacena la informacion en los bits clásicos
bell_circuits["psi+"] = qc_psi_plus # Se almacena el circuito en el diccionario

# 4. Estado |Psi-> = (|01> - |10>) / sqrt(2)
qc_psi_minus = QuantumCircuit(2, 2) # Se parte de un circuito con estado inicial |00> y dos bits clásicos para la posterior medición
qc_psi_minus.x(0) # Se empieza en |10> para que tras la H y CNOT cambie el signo
qc_psi_minus.h(0) # Se aplica la puerta H al qbit 0
qc_psi_minus.cx(0, 1) # Se aplica puerta CNOT con control 0 y target 1 
qc_psi_minus.x(1)  # Se aplica la puerta X en el qbit 1 para invertir sus valores lógicos
qc_psi_minus.measure([0, 1], [0, 1]) # Se realiza la medición de los qubits y se almacena la informacion en los bits clásicos
bell_circuits["psi-"] = qc_psi_minus # Se almacena el circuito en el diccionario
```

## 3. Simulación y Análisis Estadístico de los Estados de Bell

Una vez definidos los cuatro circuitos cuánticos, el siguiente paso consiste en ponerlos en ejecución para validar experimentalmente su comportamiento. Dado que las puertas cuánticas operan de manera determinista sobre las amplitudes de probabilidad, pero la medición es un proceso intrínsecamente probabilístico, es necesario realizar un muestreo estadístico repetido para reconstruir la distribución de probabilidad de cada estado.

Para ello, se utiliza el backend **AerSimulator** de Qiskit, un simulador de alto rendimiento que imita de forma ideal el comportamiento de un procesador cuántico. A continuación, se ejecuta cada circuito de manera independiente un total de **1024 veces** y calcula las frecuencias relativas de aparición de cada estado clásico en el momento del colapso de la función de onda.

Es importante destacar que al observar los resultados de la simulación, las probabilidades obtenidas para cada componente raramente marcarán un $50.00\%$ exacto. Esto se debe a las **fluctuaciones estadísticas del muestreo** (el llamado *error de disparo* o *shot noise*).

Matemáticamente, cada una de las 1024 ejecuciones representa un experimento aleatorio independiente. Al igual que si se lanza una moneda 1024 veces no se obtiene exactamente 512 caras y 512 cruces, el colapso de los cúbits entrelazados sigue una distribución binomial. Las frecuencias relativas experimentales se aproximarán al $50\%$ teórico a medida que el número de ejecuciones tienda a infinito.

```python
from qiskit_aer import AerSimulator

# --- Simulación y Demostración de Mediciones ---
simulator = AerSimulator()
shots = 1024

for name, qc in bell_circuits.items():
    print(f"\n--- Ejecutando Circuito para Estado {name} ---")
    job = simulator.run(qc, shots=shots) # Se ejecuta la simulación
    result = job.result()          # Se obtienen los
    counts = result.get_counts(qc) # resultados de la simulación
    
    # Calcular probabilidades experimentales
    print(f"Resultados de las mediciones (Counts): {counts}")
    for state, count in counts.items(): # Se recorren todos los estados
        prob = (count / shots) * 100 # Se calcula la probabilidad de cada estado
        print(f"  Estado |{state}>: {prob}% de probabilidad")
```
